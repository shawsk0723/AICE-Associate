# 1. AICE Associate 시험 개요 및 전체 흐름 요약

## (1) 파이프라인 쉽게 이해하기 (요리사 비유)

AI 모델링 전체 작업 흐름은 맛있는 요리를 만드는 과정과 똑같습니다.

* 1단계: 데이터 불러오기 (재료 준비)
* 재료(CSV 파일)를 요리대 위에 올리고 상태를 점검합니다.


* 2단계: 시각화/EDA (재료 상태 확인)
* 눈으로 재료의 양과 균형을 확인하듯, 그래프로 데이터의 분포를 파악합니다.


* 3단계: 전처리 - 이상치/결측치 처리 (상한 재료 다듬기)
* 상하거나 이상한 부분(이상치)을 잘라내고, 비어있는 부위(결측치)를 정리합니다.


* 4단계: 범주형 변환/원-핫 인코딩 (재료 손질 및 규격화)
* 컴퓨터는 텍스트(문자)를 읽지 못하므로 숫자로 변환(0과 1)하여 다듬습니다.


* 5단계: 데이터 분리 및 스케일링 (시식용/본요리 구분 & 크기 맞추기)
* 연습용(Train)과 테스트용(Valid)으로 나누고, 재료 크기 단위(스케일)를 일정하게 맞춥니다.


* 6단계: 머신러닝 모델링 및 평가 (기본 레시피 작성 및 평가)
* 의사결정나무(DecisionTree)나 랜덤포레스트(RandomForest) 같은 정형화된 알고리즘으로 와인 품질을 예측합니다.


* 7단계: 딥러닝 모델링 및 평가 (인공지능 특제 레시피 작성 및 평가)
* 신경망(Sequential/Dense)을 구축하여 오차를 줄여가며 최선의 예측 성능을 냅니다.

# 2. 초보자를 위한 핵심 개념 및 용어 사전

* (1) 라이브러리 임포트 및 별칭 (import ... as ...)
* 긴 이름을 가진 도구 상자를 불러올 때 별명을 붙여 편리하게 쓰는 방식입니다. (예: `import pandas as pd`, `import sklearn as sk`)




* (2) 데이터프레임 (DataFrame)과 행/열
* 엑셀 표와 같은 형태입니다. 가로줄은 '행(Row/데이터 개수)', 세로줄은 '열(Column/특성 항목)'입니다.


* (3) 이상치 (Outlier) 및 IQR (Interquartile Range)
* 정상 범위에서 크게 벗어난 이상한 값입니다. 전체 데이터 상위 75%(Q3)와 25%(Q1)의 차이인 IQR을 활용해 상한선(Upper fence)과 하한선(Lower fence)을 정하고 밖의 값을 제거합니다.




* (4) 결측치 (Null / Missing Value)와 dropna
* 비어있는(NaN) 값입니다. `dropna()`를 사용하면 비어있는 칸이 포함된 행을 깨끗하게 삭제합니다.




* (5) 원-핫 인코딩 (One-Hot Encoding, pd.get_dummies)
* 'Red', 'White' 같은 문자형 카테고리를 [1, 0], [0, 1] 형태로 바꿔주어 컴퓨터가 계산할 수 있도록 0과 1의 숫자 기둥으로 늘려주는 작업입니다.




* (6) 데이터 분리 (train_test_split)와 stratify
* 데이터를 학습용과 검증용으로 나눕니다. 이때 `stratify=y` 옵션을 주면 정답(y) 비율을 모조리 동일하게 유지하면서 똑똑하게 나눠줍니다.




* (7) 머신러닝 (DecisionTree, RandomForest, GridSearchCV)
* 스무고개처럼 정답을 찾는 의사결정나무(`DecisionTree`)와, 나무 여러 개를 합쳐 정확도를 높이는 `RandomForest`가 있습니다. `GridSearchCV`는 최고의 성능을 내는 최적의 설정값(하이퍼파라미터)을 자동으로 탐색해 줍니다.




* (8) 딥러닝 (Sequential, Dense, Dropout, EarlyStopping)
* 인공신경망 레이어를 차곡차곡 쌓아올리는 구조(`Sequential`, `Dense`)입니다. 과적합(오버피팅)을 막기 위해 일부 연결을 끄는 `Dropout`, 학습 성능이 더 올라가지 않으면 자동으로 조기 종료하는 `EarlyStopping`을 함께 활용합니다.

* **
- 대상 데이터프레임 : data
- 와이너리 평점이 upperfence 값보다 큰 이상치가 있는 행(row)을 삭제하세요
- 와이너리 평점이 lowerfence 값보다 작은 이상치가 있는 행(row)을 삭제하세요
- AI모델링에 불필요한 와인이름(E_title) 컬럼을 삭제하세요
- 전처리를 적용한 데이터프레임은 data_temp라는 변수명으로 저장하세요<br><br>

````
# (코드 셀) 코드의 빈칸을 채우고 실행하세요


q1 = data['FE_points_winery'].quantile(0.25) # 하위 25% 지점
q3 = data['FE_points_winery'].quantile(0.75) # 상위 25% 지점
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
data_temp = data.drop( data[(data['FE_points_winery'] > upper_fence) | (data['FE_points_winery'] < lower_fence)].index, axis=0)
data_temp = data_temp.drop( columns=['E_title'], axis=1)
data_temp.reset_index(drop=True, inplace=True)
````

* 핵심 키워드/함수: `drop`, `index`, `columns`

* 자주 하는 실수: 조건식 비교 연산자(`|` OR 연산) 혼동, 인덱스 초기화 누락


* 정답 코드 한 줄 해석:

````
q1 = data['FE_points_winery'].quantile(0.25)
q3 = data['FE_points_winery'].quantile(0.75) # 상위 25% 지점
print(q1, q3)
iqr = q3 - q1
print(iqr)
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr
print(lower_fence, upper_fence)
````

````
# 1단계: 상한선 초과 조건 찾기 (True/False 시리즈)
is_upper_outlier = data['FE_points_winery'] > upper_fence

# 2단계: 하한선 미만 조건 찾기 (True/False 시리즈)
is_lower_outlier = data['FE_points_winery'] < lower_fence

# 3단계: 두 조건을 합쳐 최종 이상치 조건 만들기 (OR 연산자 | 사용)
is_outlier = is_upper_outlier | is_lower_outlier

print("--- [1단계] 이상치 여부 확인 (True 개수 = 이상치 개수) ---")
print(is_outlier.value_counts())
print("\n")


# 4단계: 이상치에 해당하는 행의 '인덱스(행 번호)'만 추출하기
outlier_indices = data[is_outlier].index

print("--- [2단계] 제거할 이상치 인덱스 목록 ---")
print(outlier_indices)
print(f"제거 대상 총 개수: {len(outlier_indices)}개")
print("\n")


# 5단계: drop 함수에 인덱스 목록만 전달하여 행 제거하기
data_temp = data.drop(outlier_indices, axis=0)

print("--- [3단계] 행 제거 후 데이터 크기 변화 ---")
print(f"원래 데이터 크기: {data.shape}")
print(f"제거 후 데이터 크기: {data_temp.shape}")
print("\n")


# 6단계: 불필요한 열('E_title') 제거하기
data_temp = data_temp.drop(columns=['E_title'], axis=1)

print("--- [4단계] 열 제거 후 컬럼 목록 ---")
print(data_temp.columns)
print("\n")


# 7단계: 인덱스 재정렬하기
data_temp.reset_index(drop=True, inplace=True)

print("--- [5단계] 최종 인덱스 재정렬 확인 ---")
print(data_temp.head())
````

### **와인품질등급(Grade) 컬럼을 레이블값 y로, 나머지 컬럼을 피처값 X로 할당한 후 훈련데이터셋과 검증데이터셋으로 분리하세요.**

````
from sklearn.model_selection import train_test_split
X = data_preset.drop('Grade', axis=1)
y = data_preset['Grade']
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.3, random_state=7, stratify=y)
````

stratify=y 옵션은 데이터를 훈련용(Train)과 검증용(Valid)으로 나눌 때, 타깃 변수(y, 여기서는 'Grade')의 범주 비율을 원본 데이터와 똑같은 비율로 유지하면서 분할하라는 의미입니다.

이 개념을 '층화 추출(Stratified Sampling)'이라고 부릅니다.

왜 이 옵션이 필요할까요?

만약 전체 데이터 100개 중 와인 등급(Grade)이 다음과 같이 구성되어 있다고 가정해 보겠습니다.

A 등급: 10개 (10%)

B 등급: 90개 (90%)

이 상태에서 stratify=y 옵션 없이 무작위로 7:3으로 나누게 되면, 운이 나쁠 경우 훈련 데이터에는 A 등급이 10개 모두 들어가고, 검증 데이터에는 A 등급이 0개 들어가는 상황이 발생할 수 있습니다.

검증 데이터에 A 등급이 하나도 없다면, 모델이 A 등급을 잘 맞추는지 제대로 평가할 수 없게 됩니다.

stratify=y를 사용하면 어떻게 될까요?

y의 비율(1:9)을 보존하면서 7:3으로 정확하게 쪼개줍니다.

훈련 데이터 (70개): A 등급 7개 (10%), B 등급 63개 (90%)

검증 데이터 (30개): A 등급 3개 (10%), B 등급 27개 (90%)

요약

분류(Classification) 문제에서 특정 클래스(라벨)의 개수가 적거나 불균형할 때 필수적으로 사용하는 옵션입니다.

훈련 세트와 검증 세트 모두에 모든 클래스가 고르게 분포하도록 만들어, 모델 평가의 신뢰성을 높여줍니다.

## 2. GridSearchCV란?

**GridSearchCV**는 머신러닝 모델의 성능을 최대로 끌어올리기 위한 **최적의 하이퍼파라미터(Hyperparameter) 조합을 자동으로 찾아주는 탐색 도구**입니다.

이름의 의미를 나누어 보면 동작 원리를 쉽게 이해할 수 있습니다:

* **Grid (격자/그물망)**: 입력한 파라미터 후보들을 격자처럼 모든 경우의 수로 교차 조합합니다.
* **Search (탐색)**: 생성된 모든 조합을 하나씩 모델에 적용하여 학습을 진행합니다.
* **CV (Cross Validation, 교차 검증)**: 각 조합을 평가할 때 교차 검증을 수행하여 과적합(Overfitting)을 방지합니다.

---

## 3. 코드 파라미터 상세 설명

Random Forest 튜닝 코드를 기준으로 각 매개변수의 역할은 다음과 같습니다:

* `estimator=rf`: 탐색 대상이 될 기본 머신러닝 모델을 지정합니다.
* `param_grid=param_grid_rf`: 테스트할 하이퍼파라미터 후보 목록입니다.
  * `n_estimators`: [100, 200, 500] (3개)
  * `max_depth`: [5, 10, 20] (3개)
  * `min_samples_split`: [2, 5, 10] (3개)
  * **총 조합 수**: $3 \times 3 \times 3 = 27$가지
* `cv=5`: 5-Fold 교차 검증을 수행합니다. 데이터를 5개 조각으로 나누어 4개로 학습하고 1개로 검증하는 과정을 5회 반복합니다.
  * **총 학습 횟수**: $27 \text{조합} \times 5 \text{회} = 135$번
* `scoring="accuracy"`: 모델 평가 기준을 '정확도(Accuracy)'로 설정합니다.
* `n_jobs=-1`: 컴퓨터의 모든 CPU 코어를 병렬로 활용하여 탐색 속도를 높입니다.

* 핵심 목표: Keras Sequential 신경망 구축, EarlyStopping 적용 및 모델 학습

* 자주 하는 실수: 입력층 `input_dim=X_train.shape[1]` 누락, 라벨 원-핫 인코딩(`to_categorical`) 과정 빠뜨림


* 정답 코드 한 줄 해석:

````
import tensorflow as tf
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import Dense, Activation, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.utils import to_categorical

tf.random.set_seed(1)
````

````
# (코드 셀) 코드의 빈칸을 채우고 실행하세요

from sklearn.preprocessing import LabelEncoder

model = Sequential([
    Dense(128, activation='relu', input_dim=X_train.shape[1]),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(2, activation='sigmoid')
])

estop = EarlyStopping(monitor='val_loss', patience=14, restore_best_weights=True)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_valid = le.transform(y_valid)

y_train= to_categorical(y_train)
y_valid = to_categorical(y_valid)

history = model.fit(X_train, y_train,
                     epochs=50,
                     batch_size=128,
                     validation_data=(X_valid, y_valid),
                     callbacks=[estop])
````

* 다층 신경망을 순서대로 쌓은 후 14번 이상 성능 향상이 없으면 조기 종료하는 조건(`EarlyStopping`)을 달아 딥러닝 학습을 수행합니다.
