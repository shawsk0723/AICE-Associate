# 1. AICE Associate 시험 개요 및 전체 흐름 요약

## (1) 파이프라인 쉽게 이해하기 (요리사 비유)

AI 모델링 전체 작업 흐름은 맛있는 요리를 만드는 과정과 똑같습니다.

* 1단계: 데이터 불러오기 (재료 준비)
* 재료(CSV 파일)를 요리대 위에 올리고 상태를 점검합니다.


* 2단계: 시각화/EDA (재료 상태 확인)
* 눈으로 재료의 양과 균형을 확인하듯, 그래프로 데이터의 분포를 파악합니다.


* 3단계: 전처리 - 이상치/결측치 처리 (상한 재료 다듬기)
* 상하거나 이상한 부분(이상치)을 잘라내고, 비어있는 부위(결측치)를 정리합니다.


* 4단계: 범주형 변환/원-핫 인코딩 (재료 손질 및 규격화)
* 컴퓨터는 텍스트(문자)를 읽지 못하므로 숫자로 변환(0과 1)하여 다듬습니다.


* 5단계: 데이터 분리 및 스케일링 (시식용/본요리 구분 & 크기 맞추기)
* 연습용(Train)과 테스트용(Valid)으로 나누고, 재료 크기 단위(스케일)를 일정하게 맞춥니다.


* 6단계: 머신러닝 모델링 및 평가 (기본 레시피 작성 및 평가)
* 의사결정나무(DecisionTree)나 랜덤포레스트(RandomForest) 같은 정형화된 알고리즘으로 와인 품질을 예측합니다.


* 7단계: 딥러닝 모델링 및 평가 (인공지능 특제 레시피 작성 및 평가)
* 신경망(Sequential/Dense)을 구축하여 오차를 줄여가며 최선의 예측 성능을 냅니다.

# 2. 초보자를 위한 핵심 개념 및 용어 사전

* (1) 라이브러리 임포트 및 별칭 (import ... as ...)
* 긴 이름을 가진 도구 상자를 불러올 때 별명을 붙여 편리하게 쓰는 방식입니다. (예: `import pandas as pd`, `import sklearn as sk`)




* (2) 데이터프레임 (DataFrame)과 행/열
* 엑셀 표와 같은 형태입니다. 가로줄은 '행(Row/데이터 개수)', 세로줄은 '열(Column/특성 항목)'입니다.


* (3) 이상치 (Outlier) 및 IQR (Interquartile Range)
* 정상 범위에서 크게 벗어난 이상한 값입니다. 전체 데이터 상위 75%(Q3)와 25%(Q1)의 차이인 IQR을 활용해 상한선(Upper fence)과 하한선(Lower fence)을 정하고 밖의 값을 제거합니다.




* (4) 결측치 (Null / Missing Value)와 dropna
* 비어있는(NaN) 값입니다. `dropna()`를 사용하면 비어있는 칸이 포함된 행을 깨끗하게 삭제합니다.




* (5) 원-핫 인코딩 (One-Hot Encoding, pd.get_dummies)
* 'Red', 'White' 같은 문자형 카테고리를 [1, 0], [0, 1] 형태로 바꿔주어 컴퓨터가 계산할 수 있도록 0과 1의 숫자 기둥으로 늘려주는 작업입니다.




* (6) 데이터 분리 (train_test_split)와 stratify
* 데이터를 학습용과 검증용으로 나눕니다. 이때 `stratify=y` 옵션을 주면 정답(y) 비율을 모조리 동일하게 유지하면서 똑똑하게 나눠줍니다.




* (7) 머신러닝 (DecisionTree, RandomForest, GridSearchCV)
* 스무고개처럼 정답을 찾는 의사결정나무(`DecisionTree`)와, 나무 여러 개를 합쳐 정확도를 높이는 `RandomForest`가 있습니다. `GridSearchCV`는 최고의 성능을 내는 최적의 설정값(하이퍼파라미터)을 자동으로 탐색해 줍니다.




* (8) 딥러닝 (Sequential, Dense, Dropout, EarlyStopping)
* 인공신경망 레이어를 차곡차곡 쌓아올리는 구조(`Sequential`, `Dense`)입니다. 과적합(오버피팅)을 막기 위해 일부 연결을 끄는 `Dropout`, 학습 성능이 더 올라가지 않으면 자동으로 조기 종료하는 `EarlyStopping`을 함께 활용합니다.

### 10번 문항

* 핵심 목표: DecisionTree 및 RandomForest 모델을 GridSearchCV로 하이퍼파라미터 튜닝


* 핵심 키워드/함수: `GridSearchCV`, `DecisionTreeClassifier`, `RandomForestClassifier`, `fit`

* 자주 하는 실수: `param_grid` 딕셔너리 내 키 값 오탈자(예: `max_depth` 오타), `random_state` 누락


* 정답 코드 한 줄 해석:

```
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# Decision Tree 튜닝
dt = DecisionTreeClassifier(random_state=7)
param_grid_dt = {"max_depth": [3, 5, 7, 10], "min_samples_split": [2, 3, 5, 10]}
gs_dt = GridSearchCV(estimator=dt, param_grid=param_grid_dt, cv=5, scoring="accuracy", n_jobs=-1)
gs_dt.fit(X_train, y_train)

# Random Forest 튜닝
rf = RandomForestClassifier(random_state=7)
param_grid_rf = {"n_estimators": [100, 200, 500], "max_depth": [5, 10, 20], "min_samples_split": [2, 5, 10]}
gs_rf = GridSearchCV(estimator=rf, param_grid=param_grid_rf, cv=5, scoring="accuracy", n_jobs=-1)
gs_rf.fit(X_train, y_train)
```

* 후보 조건들 중 교차검증(cv=5)을 거쳐 가장 우수한 설정값을 찾도록 교차 검증 및 학습을 실행합니다.

* 핵심 키워드/함수: `Sequential`, `Dense`, `Dropout`, `EarlyStopping`, `model.compile`, `model.fit`

* 자주 하는 실수: 입력층 `input_dim=X_train.shape[1]` 누락, 라벨 원-핫 인코딩(`to_categorical`) 과정 빠뜨림


* 정답 코드 한 줄 해석:

```
model = Sequential([
    Dense(128, activation='relu', input_dim=X_train.shape[1]),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(2, activation='sigmoid')
])
estop = EarlyStopping(monitor='val_loss', patience=14, restore_best_weights=True)
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# 라벨 인코딩 처리 후
history = model.fit(X_train, y_train, epochs=50, batch_size=128, validation_data=(X_valid, y_valid), callbacks=[estop])
```

* 다층 신경망을 순서대로 쌓은 후 14번 이상 성능 향상이 없으면 조기 종료하는 조건(`EarlyStopping`)을 달아 딥러닝 학습을 수행합니다.

## (1) 코드 암기 패턴 (틀로 기억하기)

* 머신러닝 모델링 4단계 공식:
1. 불러오기: `from sklearn... import ...`

2. 선언하기: `model = 모델명()`

3. 학습하기: `model.fit(X_train, y_train)`

4. 예측 및 평가: `pred = model.predict(X_valid)` → `accuracy_score(y_valid, pred)`



* 딥러닝 모델링 4단계 공식:
1. 모델 선언: `model = Sequential([ ... ])`

2. 컴파일: `model.compile(optimizer='adam', loss='...', metrics=['accuracy'])`

3. 조기종료: `estop = EarlyStopping(monitor='val_loss', patience=...)`

4. 학습: `history = model.fit(..., callbacks=[estop])`




## (2) 자주 발생하는 파이썬 에러 정정 팁

* SyntaxError (문법 에러):
* 괄호 `()`, `[]`의 짝이 맞지 않거나 따옴표를 닫지 않았을 때 발생합니다. 코드 줄 끝부분과 오타를 확인하세요.


* KeyError (키 에러):
* 데이터프레임 컬럼명이 틀렸을 때 주로 발생합니다. 대소문자나 철자를 오타 없이 썼는지 반드시 확인하세요. (예: `E_FE_wine_kind` 대소문자 구분)




* ValueError (값 에러):
* 행과 열의 모양(Shape)이 맞지 않거나 데이터 유형이 맞지 않을 때 발생합니다. 스케일링이나 인코딩 과정에서 변수가 잘 변환되었는지 `info()`나 `shape`로 점검하세요.





## (3) 시험 시간 배분 전략 (전처리 50% / 모델링 50%)

* 데이터 전처리 (1번 ~ 9번):

* 모델링 및 평가 (10번 ~ 14번):
* 패턴이 정형화되어 있으므로 문제지 가이드의 조건(파라미터 숫자, 변수명)을 빠짐없이 대입하는 데 집중하여 남은 시간(50%)을 활용하세요.
